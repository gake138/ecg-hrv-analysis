```python
import numpy as np
import scipy.signal as signal
import matplotlib
matplotlib.rcParams['font.sans-serif'] = ['Microsoft YaHei']  # 中文显示
matplotlib.rcParams['axes.unicode_minus'] = False             # 负号显示
import matplotlib.pyplot as plt
import wfdb

DATA_PATH = r'D:\mitdb'   # 数据集位置
REC       = '100'         # 当前分析的记录号，以后换记录只改这里

record = wfdb.rdrecord(rf'{DATA_PATH}\{REC}')
ann    = wfdb.rdann(rf'{DATA_PATH}\{REC}', 'atr')

ecg   = record.p_signal[:, 0]   # 导联0的原始信号
fs    = record.fs               # 采样率 360Hz
r_ref = ann.sample              # 专家标注的心跳位置

print(f'记录{REC}: 采样率{fs}Hz, 长度{len(ecg)}点, 约{len(ecg)/fs/60:.1f}分钟, 标注心跳{len(r_ref)}个')
```

```python
def design_filters(fs):
    b_hp, a_hp       = signal.butter(2, 0.5, btype='highpass', fs=fs)  # 去基线漂移
    b_notch, a_notch = signal.iirnotch(50, Q=30, fs=fs)                # 去50Hz工频
    b_lp, a_lp       = signal.butter(2, 40, btype='lowpass', fs=fs)    # 去肌电高频
    return (b_hp, a_hp), (b_notch, a_notch), (b_lp, a_lp)

def preprocess_ecg(ecg, fs):
    (b_hp, a_hp), (b_notch, a_notch), (b_lp, a_lp) = design_filters(fs)
    x = signal.filtfilt(b_hp,    a_hp,    ecg)   # filtfilt=零相位，不移动波形位置
    x = signal.filtfilt(b_notch, a_notch, x)
    x = signal.filtfilt(b_lp,    a_lp,    x)
    return x

ecg_filt = preprocess_ecg(ecg, fs)
print('滤波完成，长度:', len(ecg_filt))
```

```python
n = int(10 * fs)
t = np.arange(n) / fs

fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
axes[0].plot(t, ecg[:n], color='gray');        axes[0].set_title('原始 Raw')
axes[1].plot(t, ecg_filt[:n], color='tab:red'); axes[1].set_title('滤波后 Filtered')
axes[1].set_xlabel('Time (s)')
plt.tight_layout(); plt.show()
```

```python
i   = r_ref[10]                              # 以第11个标注心跳为中心
s, e = i - int(0.3*fs), i + int(0.4*fs)      # 前0.3s后0.4s = 一个完整心动周期
tt  = np.arange(s, e) / fs

plt.figure(figsize=(10, 4))
plt.plot(tt, ecg[s:e],      color='gray',    label='raw')
plt.plot(tt, ecg_filt[s:e], color='tab:red', label='filtered')
plt.legend(); plt.xlabel('Time (s)'); plt.title('单拍放大：红灰R波应重合')
plt.show()
```

```python
from scipy.signal import welch

f1, p1 = welch(ecg,      fs, nperseg=2048)
f2, p2 = welch(ecg_filt, fs, nperseg=2048)

plt.figure(figsize=(10, 4))
plt.semilogy(f1, p1, color='gray',    label='raw')
plt.semilogy(f2, p2, color='tab:red', label='filtered')
plt.xlim(0, 100); plt.axvline(50, color='k', ls='--', alpha=.3)
plt.xlabel('Frequency (Hz)'); plt.ylabel('PSD'); plt.legend()
plt.title('功率谱：0.5以下/40以上被压低，50Hz有凹陷')
plt.show()
```

```python
from scipy.signal import find_peaks

def pan_tompkins_detect(ecg, fs):
    # ① 带通5-15Hz：只留QRS能量，压掉P/T波
    b, a = signal.butter(2, [5, 15], btype='bandpass', fs=fs)
    x = signal.filtfilt(b, a, ecg)
    # ② 五点微分：突出QRS的陡峭斜率
    h = np.array([-1, -2, 0, 2, 1]) * fs / 8
    y = np.convolve(x, h, mode='same')
    # ③ 平方：全变正 + 拉大与噪声的对比度
    y = y ** 2
    # ④ 150ms滑窗积分：每个QRS聚成一个孤立鼓包
    w = int(0.15 * fs)
    y = np.convolve(y, np.ones(w)/w, mode='same')
    # ⑤ 阈值检测：数据驱动阈值 + 200ms不应期(生理极限心率300bpm)
    thr = 0.3 * np.percentile(y, 99)
    bumps, _ = find_peaks(y, height=thr, distance=int(0.2*fs))
    # 回映射：回到干净波形±50ms内找真实R波顶点(积分鼓包中心≠R顶点)
    half = int(0.05 * fs)
    r_peaks = np.array([p - half + np.argmax(ecg_filt[p-half:p+half+1])
                        for p in bumps])
    return r_peaks, x, y

r_peaks, x_bp, y_int = pan_tompkins_detect(ecg_filt, fs)
print(f'检测到 {len(r_peaks)} 个心跳 (标注为 {len(r_ref)} 个)')
```

```python
h      = np.array([-1, -2, 0, 2, 1]) * fs / 8
y_diff = np.convolve(x_bp, h, mode='same')
y_sq   = y_diff ** 2

fig, axes = plt.subplots(4, 1, figsize=(12, 9), sharex=True)
n = int(10 * fs); t = np.arange(n) / fs
axes[0].plot(t, x_bp[:n]);                  axes[0].set_title('① 带通5-15Hz后：P/T波已减弱')
axes[1].plot(t, y_diff[:n]);                axes[1].set_title('② 微分：平缓段归零，只剩QRS陡坡')
axes[2].plot(t, y_sq[:n]);                  axes[2].set_title('③ 平方：全为正，对比度拉满')
axes[3].plot(t, y_int[:n], color='tab:red'); axes[3].set_title('④ 积分：每拍一个孤立鼓包')
axes[3].set_xlabel('Time (s)')
plt.tight_layout(); plt.show()
```

```python
# 只保留"心跳类"标注：~是噪声标记、+是节律标记，不算心跳
BEAT = set('NLRBBAaJSVrFejnE/fQ')
ref  = ann.sample[[s in BEAT for s in ann.symbol]]

def evaluate(det, ref, fs, tol_ms=150):      # 150ms是AAMI标准的匹配容差
    tol = int(tol_ms/1000 * fs)
    matched = np.zeros(len(ref), bool)
    tp = fp = 0; errs = []
    for p in det:
        j = np.argmin(np.abs(ref - p))       # 找最近的真值心跳
        if abs(ref[j] - p) <= tol and not matched[j]:
            tp += 1; matched[j] = True       # 命中，且该真值只能被用一次
            errs.append((p - ref[j])/fs*1000)
        else:
            fp += 1                          # 匹配不上 = 误检
    fn = int((~matched).sum())               # 没被匹配到的真值 = 漏检
    se  = tp/(tp+fn) * 100
    ppv = tp/(tp+fp) * 100
    return tp, fp, fn, se, ppv, np.array(errs)

tp, fp, fn, se, ppv, errs = evaluate(r_peaks, ref, fs)
print(f'TP={tp}  FP={fp}  FN={fn}')
print(f'灵敏度 Se    = {se:.2f}%   (目标≥98%)')
print(f'阳性预测 +P  = {ppv:.2f}%   (目标≥98%)')
print(f'定位误差: 均值 {errs.mean():+.1f}ms, 标准差 {errs.std():.1f}ms')
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.hist(errs, bins=40)
plt.axvline(0, color='k', ls='--')
plt.xlabel('误差 (ms)'); plt.ylabel('次数'); plt.title('定位误差分布：应以0为中心窄峰')

plt.subplot(1, 2, 2)
n = int(10*fs); t = np.arange(n)/fs
plt.plot(t, ecg_filt[:n], label='ECG')
m = r_peaks[r_peaks < n]
plt.plot(m/fs, ecg_filt[m], 'rv', label='检测R波')
plt.xlabel('Time (s)'); plt.legend(); plt.title('前10秒检测结果')

plt.tight_layout(); plt.show()
```

```python
rr = np.diff(r_peaks) / fs * 1000        # 相邻R波间隔 → 毫秒(HRV的标准单位)
beat_time = r_peaks[1:] / fs             # 每个RR对应的时刻(秒)

# 清洗：剔除生理不可能值 + 远离中位数的离群拍(如室早后代偿间歇)
ok = (rr > 300) & (rr < 2000)
ok &= np.abs(rr - np.median(rr)) < 0.2 * np.median(rr)
rr_clean, t_clean = rr[ok], beat_time[ok]
print(f'RR {len(rr)}个 → 清洗后 {len(rr_clean)}个, 平均心率 {60000/rr_clean.mean():.1f} bpm')
```

```python
sdnn  = rr_clean.std()                      # 整体变异性
diffs = np.diff(rr_clean)
rmssd = np.sqrt((diffs**2).mean())          # 逐拍波动 → 副交感活性
pnn50 = (np.abs(diffs) > 50).mean() * 100   # 相邻差>50ms的比例

print(f'平均RR = {rr_clean.mean():.1f} ms (心率 {60000/rr_clean.mean():.1f} bpm)')
print(f'SDNN  = {sdnn:.1f} ms')
print(f'RMSSD = {rmssd:.1f} ms')
print(f'pNN50 = {pnn50:.1f} %')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(t_clean, rr_clean, lw=.5)
axes[0].set_xlabel('Time (s)'); axes[0].set_ylabel('RR (ms)')
axes[0].set_title('心动周期序列 Tachogram')

axes[1].plot(rr_clean[:-1], rr_clean[1:], '.', ms=2, alpha=.3)
axes[1].set_xlabel('RR$_n$ (ms)'); axes[1].set_ylabel('RR$_{n+1}$ (ms)')
axes[1].set_title('Poincaré图：沿对角线的椭圆')
axes[1].axis('equal')
plt.tight_layout(); plt.show()
sd1 = np.std(rr_clean[:-1] - rr_clean[1:]) / np.sqrt(2)
sd2 = np.std(rr_clean[:-1] + rr_clean[1:]) / np.sqrt(2)
print(f'SD1 = {sd1:.1f} ms (≈副交感), SD2 = {sd2:.1f} ms (≈整体慢变异)')
```

```python
# RR序列是非均匀采样(每拍一个点) → 必须插值到4Hz均匀网格才能做FFT
fs_hr = 4
t_grid   = np.arange(t_clean[0], t_clean[-1], 1/fs_hr)
rr_interp = np.interp(t_grid, t_clean, rr_clean)
rr_interp = rr_interp - rr_interp.mean()     # 去均值

f, p = welch(rr_interp, fs_hr, nperseg=1024) # 窗约4分钟, 分辨率≈0.004Hz
lf = p[(f>=0.04) & (f<0.15)].sum()
hf = p[(f>=0.15) & (f<0.40)].sum()

plt.figure(figsize=(10, 4))
plt.semilogy(f, p)
plt.axvspan(0.04, 0.15, alpha=.2, color='orange', label=f'LF={lf:.0f}')
plt.axvspan(0.15, 0.40, alpha=.2, color='green',  label=f'HF={hf:.0f}')
plt.xlim(0, 0.5); plt.xlabel('Frequency (Hz)')
plt.ylabel('PSD (ms²/Hz)'); plt.title('HRV频谱'); plt.legend()
plt.tight_layout(); plt.show()
print(f'LF = {lf:.0f} ms², HF = {hf:.0f} ms², LF/HF = {lf/hf:.2f}')
```

```python
# ============ T4 验收数据快照 ============
sdnn  = rr_clean.std()
diffs = np.diff(rr_clean)
rmssd = np.sqrt((diffs**2).mean())
pnn50 = (np.abs(diffs) > 50).mean() * 100
lf, hf = 31140, 122220      # 用你实测值

print('========== ECG项目最终指标汇总 ==========')
print(f'【T3 检测】Se=100%, +P=100%, 误差={errs.mean():+.1f}±{errs.std():.1f} ms')
print(f'【T4 时域】平均RR={rr_clean.mean():.0f}ms ({60000/rr_clean.mean():.0f}bpm)')
print(f'          SDNN={sdnn:.1f}ms  RMSSD={rmssd:.1f}ms  pNN50={pnn50:.1f}%')
print(f'【T4 频域】LF={lf:.0f}ms²  HF={hf:.0f}ms²  LF/HF={lf/hf:.2f}')
print(f'          Poincaré: SD1={np.std(rr_clean[:-1]-rr_clean[1:])/np.sqrt(2):.1f}ms, '
      f'SD2={np.std(rr_clean[:-1]+rr_clean[1:])/np.sqrt(2):.1f}ms')
```
